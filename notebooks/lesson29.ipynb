Streamlit
- Languages: Python
- Supports all of the main Python plotting libraries, including Matplotlib, Seaborn, Plotly, etc.
- Open source
- No web development knowledge needed
- Easy to use, less code and easily manageable
- No support for Jupyter Notebook
- can easily be deployed and supports almost all the deployment platforms


Panel
- Languages: Python only 
- Supports all of the main Python plotting libraries, including Matplotlib, Seaborn, Plotly, etc.
- Open source
- Supports multipage applications
- No web development knowledge needed
- Great design flexibility
- Supports Jupyter Notebook
- can easily be deployed and supports almost all the deployment platforms

Voila
- Languages: Python, C++, Julia
- Supports all of the main Python plotting libraries, including Matplotlib, Seaborn, Plotly, etc.
- Open source
- No web development knowledge needed
- Limited design flexibility
- Exceptional Jupyter support
- Can easily be deployed and supports almost all the deployment platforms

Dash
- Language: Python, R, Julia
- Primarily built for use with the Python library Plotly
- Supports multipage applications
- Open source
- Need basic knowledge of HTML
- Easy to use create basic dashboard applications
- Main strength is its incredible design flexibility
- No support for Jupyter Notebook
- Can easily be deployed and supports almost all the deployment platforms.
