Regression Models
* Linear regression is a statistical machine learning method you can use to quantify and make predictions based on relationships between numerical variables
- Simple linear regression - one predictor, one predictant.
- Multiple linear regression - multiple predictors, one predictant.

Artificial Intelligence: A machine or application with the capacity to autonomously execute upon predictions it makes from data 
- Two elements:
1. Prediction: predictive modeling (from data science)
2. Execution: autonomous response (from engineering)

Machine Learning
* The process of deploying compute power to identify trends and patterns in data in order to make predictions about how that data will behave in the future

With respect to AI, the decision engine in a piece of AI technology can be comprised of machine learning models.|

Robotics: The design, construction, and operation of programmable machines that are capable of automatically executing a series of complex tasks
