Machine Learning
- The process of deploying compute power to identify trends and patterns in data in order to make predictions about how that data will behave in the future.

* There are hundreds (if not thoudsands) of machine learning use

- Sales forecasting
- Customer segment analysis
- Insurance claim fraud detection
- Hedge fund classification

What Is Machine Learning?
* The models can be used to:
- Simplifgy inputs by mapping them into a lower dimensional space (dimensionality reduction)
- Group similar data points together without knowing the groups beforehand (density estimation)

Machine Learning Lingo
- Feature: a feature is synonymous with the terms variable, column, attribute, and field.
- instance: an instance is synonymous with the terms row, observation, data point, value, and case.
- Target: a target variable is synonymous with the terms predicted and dependent variable.
- Data: the term data is used synonymously to refer to a predictor or set of predictor variables.

Machine Learning Test and Training Sets
- Use random sampling to generate random samples and then break the data into test and training sets:
* 4/5 of your dataset: training set (use it to train your model)
* 1/5 of your dataset: test set (for testing to see how well the model performs)

Types of Machine Learning Methods
* Supervised Learning: Making predictions straight from labeled data.
* Unsurpervised Learning: Making predictions straight from unlabeled data.
* Semi-Supervised Learning: Uses both labeled and unlabeled data to make a set of predictions
